### 1-Encapsulation 

```python
class Student():
    def __init__(self,name,roll_no,course,compus,age,email):
        # Private Variable/attributes
        self.__name= name
        self.__roll_no =roll_no 
        self.__course= course
        self.__compus=compus
        self.__age=age
        self.__email=email
        self.__gender="male"
    def Students_info(self):
        print(f"""
        name = {self.name}
        Roll_no = {self.roll_no}
        course = {self.course}
        Compus = {self.compus}
        Age = {self.age}
        Email = {self.email}
        """)
    def Attended_class(self):
        print(f"{self.__name} is attended class")
    def Appear_Exam(self):
        print(f"{self.__roll_no} is appearing in exam")
    def Get_name(self):
        return self.__name
    def Re_name(self,name):
        lst=["john","Robery","Humaish"]
        if name not in lst:
            self.__name=name
        else:
            return f"{name} is not allowed"
```

```python
class Student():
    def __init__(self,name,roll_no,course,compus,age,email):
        # Protected Variable/attributes
        self._name= name
        self._roll_no =roll_no 
        self._course= course
        self._compus=compus
        self._age=age
        self._email=email
        self._gender="male"
    def Students_info(self):
        print(f"""
        name = {self._name}
        Roll_no = {self._roll_no}
        course = {self._course}
        Compus = {self._compus}
        Age = {self._age}
        Email = {self._email}
        """)
    def Attended_class(self):
        print(f"{self._name} is attended class")
    def Appear_Exam(self):
        print(f"{self._roll_no} is appearing in exam")
    def Get_name(self):
        return self._name
    def Re_name(self,name):
        lst=["john","Robery","Humaish"]
        if name not in lst:
            self._name=name
        else:
            return f"{name} is not allowed"
```

### 2-Inheritance

```python
class father():
    def __init__(self,Eyes,Hair,Height):
        self.Eyes=Eyes
        self.Hair=Hair
        self.Height=Height
    def f_job(self):
        print("You Have doing Jobs")
```

```python
f1 = father("Blue","Brown",5.5)
```

```python
f1.Hair
```

```python
class sons(father):
    def __init__(self,Eyes,Hair,Height,Name,Age):
        super().__init__(Eyes,Hair,Height)
        self.Name=Name
        self.Age=Age
    def f_job(self):
        return "You do not Have doing Jobs Beacause underage"
```

```python
s1 = sons("Green","yellow",4.4,"john",5)
```

```python
s1.Age
```

```python
f1.Eyes
```

```python
s1.Eyes
```

### 4-Abstraction

```python
class car():
    def __init__(self):
        self.acc=False
        self.brake=False
        self.cluth=False
    def start(self):
        self.acc=True
        self.cluth=True
        print("Car Started")
```
